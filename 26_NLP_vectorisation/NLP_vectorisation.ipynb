# NLP - Vectorisation

```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
# text = "Natural Language Processing (NLP) is a branch of Artificial Intelligence (AI) that focuses on enabling computers to understand, interpret, and generate human language. "

corpus = [
    "NLP is amazing and helps computers understand language.",
    "I love working with natural language processing.",
    "Understanding language is essential for communication."
]
```

```python
corpus
```

# Vectorisation - Using TF-IDF

```python
processed_corpus = []

for text in corpus:
    doc = nlp(text)  # Step 1: Process text with spaCy
    tokens = []      # Step 2: Collect valid tokens
    for token in doc:
        if not token.is_stop and token.is_alpha:  # Filter stopwords and punctuations
            lemma = token.lemma_.lower()          # Lemmatize and lowercase
            tokens.append(lemma)
    cleaned_text = " ".join(tokens)               # Step 3: Join back into a cleaned sentence
    processed_corpus.append(cleaned_text)
```

```python
processed_corpus
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(processed_corpus)

# print('vect', vectorizer.get_feature_names_out())
# print('X', X.toarray())

import pandas as pd
df = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
print(df)
```

# Vectorisation - using Bag of Words

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer_bw = CountVectorizer()
X_bw = vectorizer_bw.fit_transform(processed_corpus)

# print(vectorizer_bw.get_feature_names_out())
# print(X_bw.toarray())

df_bw = pd.DataFrame(X_bw.toarray(), columns=vectorizer_bw.get_feature_names_out())
print(df_bw)
```
